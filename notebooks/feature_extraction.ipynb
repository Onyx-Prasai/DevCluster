### Opened pull requests

```python
# Compute pull_requests_opened, issues_opened, and readme_presence_ratio
user_login = data.get("user_data", {}).get("login")

pull_requests_opened = 0
per_repo_prs_opened = {}
for repo_name, prs in data.get("repos_pull_requests", {}).items():
    count = 0
    for pr in prs or []:
        if pr.get("user", {}).get("login") == user_login:
            count += 1
    if count:
        per_repo_prs_opened[repo_name] = count
    pull_requests_opened += count

print("pull_requests_opened:", pull_requests_opened)
print("per_repo_prs_opened (non-zero):", per_repo_prs_opened)

issues_opened = 0
per_repo_issues_opened = {}
for repo_name, issues in data.get("repos_issues", {}).items():
    count = 0
    for issue in issues or []:
        if issue.get("pull_request"):
            continue
        if issue.get("user", {}).get("login") == user_login:
            count += 1
    if count:
        per_repo_issues_opened[repo_name] = count
    issues_opened += count

print("issues_opened:", issues_opened)
print("per_repo_issues_opened (non-zero):", per_repo_issues_opened)

readme_count = 0
for repo_name, contents in data.get("repos_contents", {}).items():
    contents = contents or []
    has_readme = any((item.get("type") == "file" and item.get("name", "").lower().startswith("readme")) for item in contents)
    if has_readme:
        readme_count += 1

public_repo_count = len(data.get("repos_data", []))
readme_presence_ratio = (readme_count / public_repo_count) if public_repo_count else 0.0

print("readme_count:", readme_count)
print("public_repo_count:", public_repo_count)
print("readme_presence_ratio:", readme_presence_ratio)
```
